```python
import pandas as pd
```

```python
df = pd.read_csv("data/winemag-data-130k-v2.csv", index_col=0)
```

```python
df.head()
```

```python
df.columns
```

```python
df["points"].value_counts(ascending=False)
```

```python
df.groupby("points")["price"].min()
```

```python
df.groupby(["country", "province"]).apply(
    lambda x: x.loc[x["points"].idxmax()], include_groups=False
)
```

```python
df.groupby("country")["price"].agg([len, "min", "max"]).head(8)
```

The `len` method gives the number of entries for each country

```python
# length of written description for the specific wine produced in the specified country, province
countries_reviews = df.groupby(["country", "province"])["description"].agg(len)
countries_reviews
```

```python
type(countries_reviews.index)
```

```python
countries_reviews = countries_reviews.reset_index()
countries_reviews
```

```python
countries_reviews.sort_values(by="description")
```

```python
countries_reviews.sort_values(by="description", ascending=False)
```

```python
countries_reviews.sort_values(["country", "description"]).loc[countries_reviews["country"] == "Italy"]
```
