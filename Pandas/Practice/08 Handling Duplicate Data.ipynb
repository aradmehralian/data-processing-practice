## Handling Duplicate Data in Pandas

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(
    {
        "Name": ["John", "Jane", "John", "Mike", "Jane"],
        "Age": [28, 32, 28, 45, 32],
        "City": ["New York City", "Boston", "New York City", "Chicago", "Denver"],
    }
)

df
```

```python
print(f"Mean of age with duplicates:\n{df['Age'].mean()}")
```

```python
print("Mean of age without duplicates:")
print(df.drop_duplicates()["Age"].mean())
```

### Removing Duplicate Data

```python
df.drop_duplicates(subset=["Name"], keep="last")
```

```python
df.drop_duplicates(subset=["Name", "Age"], ignore_index=True)
```

```python
df = pd.DataFrame({"A": [1, 2, 2, 3], "B": ["a", "b", "b", "c"], "C": [10, 20, 20, 30]})

df
```

```python
df.duplicated()
```

```python
df.duplicated().any()
```

```python
df.drop_duplicates(keep="first")  # by default keeps first duplicate
```

```python
df = pd.DataFrame(
    {
        "Name": ["John", "Jane", "John", "Mike", "Jane"],
        "Age": [28, 32, 28, 45, 33],
        "City": ["New York", "Boston", "Chicago", "Chicago", "Boston"],
    }
)

df
```

```python
df.duplicated(subset=["Name", "Age"]).any()
```

```python
df.drop_duplicates(subset=["Name", "Age"], keep="last", inplace=True, ignore_index=True)
```

```python
df
```

```python
df.isnull().any()
```
