### Concatenating DataFrames

```python
customers_us = pd.DataFrame(
    {
        "customer_id": [1, 2, 3],
        "name": ["John Doe", "Jane Smith", "Bob Johnson"],
        "country": ["USA", "USA", "USA"],
    }
)
```

```python
customers_us
```

```python
customers_uk = pd.DataFrame(
    {
        "customer_id": [4, 5, 6],
        "name": ["Alice Brown", "Charlie Davis", "Emma Wilson"],
        "country": ["UK", "UK", "UK"],
    }
)

customers_uk
```

```python
pd.concat([customers_us, customers_uk])
```

```python
customer_info = pd.DataFrame(
    {
        "customer_id": [1, 2, 3, 4],
        "name": ["John Doe", "Jane Smith", "Bob Johnson", "Alice Brown"],
    }
)

customer_info
```

```python
customer_orders = pd.DataFrame(
    {
        "customer_id": [1, 2, 3, 4],
        "total_orders": [5, 3, 7, 2],
        "total_spend": [500, 300, 700, 200],
    }
)

customer_orders
```

```python
pd.concat([customer_info, customer_orders], axis=1)
```

```python
customer_orders = customer_orders.rename(
    index=dict(zip(customer_orders.index, list("ABCD")))
)
customer_orders
```

```python
customer_orders
```

```python
pd.concat([customer_info, customer_orders], axis=1)
```

```python
pd.concat([customers_us, customers_uk]).reset_index(drop=True)
```

```python
# hierarchial index
df = pd.concat([customers_us, customers_uk], axis=0, keys=["US", "UK"])
df
```

```python
df.loc["US"]
```

```python
df.loc[("UK", 0), "name"]
```

```python
pd.concat([customers_us, customers_uk], ignore_index=True)
```

```python
customers_us["state"] = ["NY", "CA", "TX"]
customers_us
```

```python
pd.concat([customers_us, customers_uk], join="inner", ignore_index=True)
```

```python
pd.concat([customers_us, customers_uk], join="outer", ignore_index=True)
```

```python
customers_uk["loyalty_score"] = [100, 150, 75]
customers_uk
```

```python
pd.concat([customers_us, customers_uk])
```

### Merging DataFrames

```python
customers
```

```python
orders
```

```python
# inner merge
pd.merge(customers, orders, on="customer_id", how="inner")
```

```python
# outer merge
pd.merge(customers, orders, on="customer_id", how="outer")
```

```python
pd.merge(customers, orders, how="cross")
```

```python
# left merge
pd.merge(customers, orders, on="customer_id", how="left")
```

```python
# right merge
pd.merge(customers, orders, on="customer_id", how="right")
```

```python
customers_indexed = customers.set_index("customer_id")
customers_indexed
```

```python
customers_indexed["name"]
```

```python
orders_indexed = orders.set_index("customer_id")
orders_indexed
```

```python
# merge on index
pd.merge(customers_indexed, orders_indexed, left_index=True, right_index=True)
```

### Joining DataFrames

```python
customers = pd.DataFrame(
    {
        "customer_id": [101, 102, 103, 104, 105],
        "name": [
            "Alice Brown",
            "Bob Charles",
            "Charlie Davis",
            "David Evans",
            "Eve Franklin",
        ],
        "email": [
            "alice@example.com",
            "bob@example.com",
            "charlie@example.com",
            "david@example.com",
            "eve@example.com",
        ],
    }
)
```

```python
customers.set_index("customer_id", inplace=True)
```

```python
customers
```

```python
orders = pd.DataFrame(
    {
        "order_id": [1001, 1002, 1003, 1004, 1005],
        "customer_id": [101, 103, 102, 105, 101],
        "order_date": [
            "2023-06-01",
            "2023-06-02",
            "2023-06-03",
            "2023-06-04",
            "2023-06-05",
        ],
        "total_amount": [150.50, 200.75, 50.25, 300.00, 175.80],
    }
).set_index("customer_id")
```

```python
orders
```

```python
supports_tickets = pd.DataFrame(
    {
        "ticket_id": [5001, 5002, 5003, 5004],
        "customer_id": [102, 104, 101, 103],
        "issue": [
            "Payment failed",
            "Item not received",
            "Wrong item sent",
            "Refund request",
        ],
        "status": ["Resolved", "In Progress", "Resolved", "Pending"],
    }
).set_index("customer_id")
```

```python
supports_tickets
```

```python
pd.merge(customers, orders, left_index=True, right_index=True, how="outer")
```

```python
customers.join(orders, how="outer")
```

```python
# inner join
customers.join(orders, how="inner")
```

```python
# left join
customers.join(orders, how="left")
```

```python
# right join
customers.join(orders, how="right")
```

```python
customers.join([orders, supports_tickets], how="outer")
```

```python
customers["rating"] = [4.5, 3.8, 4.2, 5.0, 4.7]
customers
```

```python
orders["rating"] = [4.0, 4.5, 3.5, 5.0, 4.8]
orders
```

```python
customers.join(orders, lsuffix="_customer", rsuffix="_order", how="inner")
```
