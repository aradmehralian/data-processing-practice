## Reshaping and Pivot Tables

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(25)
```

```python
products = ["Laptop", "Smartphone", "Tablet", "Smartwatch"]
regions = ["North", "South", "East", "West"]
dates = pd.date_range(start="2024-01-01", end="2024-12-31", freq="ME")
```

```python
data = []
for date in dates:
    for product in products:
        for region in regions:
            if product == 'Laptop':
                sales = np.random.randint(800, 2000)
                units = np.random.randint(5, 20)
            elif product == 'Smartphone':
                sales = np.random.randint(500, 1200)
                units = np.random.randint(10, 30)
            elif product == 'Tablet':
                sales = np.random.randint(300, 800)
                units = np.random.randint(5, 15)
            else:  # Smartwatch
                sales = np.random.randint(200, 500)
                units = np.random.randint(8, 25)

            if date.month in [11, 12]:  # Holiday season
                sales = int(sales * 1.2)
                units = int(units * 1.2)

            data.append([date, product, region, sales, units])
```

```python
df = pd.DataFrame(data, columns=["Date", "Product", "Region", "Sales", "Units"])
```

```python
df["Average Price"] = (df["Sales"] / df["Units"]).round(decimals=2)
```

### Pivoting Data

```python
data = {
    "variable": ["A"] * 3 + ["B"] * 3 + ["C"] * 3 + ["D"] * 3,
    "value": range(12),
    "date": pd.to_datetime(["2025-06-12", "2025-08-04", "2024-11-23"] * 4)
}

df_example = pd.DataFrame(data)
df_example
```

```python
# using pivot()
df_example.pivot(index="date", columns="variable", values="value")
```

```python
df
```

```python
df.pivot_table(index="Product", columns="Region", values="Sales", aggfunc="sum")
```

```python
df.pivot_table(
    index=["Region", "Product"],
    columns="Date",
    values=["Sales", "Units"],
    aggfunc={"Sales": "sum", "Units": "mean"},
    fill_value=0
)
```

```python
df.pivot_table(
    index="Product",
    values=["Sales", "Units", "Average Price"],
    aggfunc={"Sales": ["mean", "sum"], "Units": ["mean", "sum"], "Average Price": ["mean", "min", "max"]}
)
```
