```python
df = pd.DataFrame({"Size": sizes, "Count": [5, 2, 8, 3, 4]})
```

```python
df
```

```python
df["Size"].cat.categories
```

```python
df.sort_values("Size")
```

```python
df.sort_values("Size", ascending=False)
```

```python
df = pd.DataFrame(
    {
        "Color": pd.Categorical(["Red", "Blue", "Green", "Blue", "Red"]),
        "Value": np.array([10, 20, 15, 25, 5], dtype="u1"),
    }
)
```

### Memory Efficiency of Categorical Data

```python
n = 1_000_000
df = pd.DataFrame({"Category_str": np.random.choice(list("ABCDE"), size=n)})
```

```python
df
```

```python
df.memory_usage(deep=True, index=False)
```

```python
58000000 / 1e6  # memory consumption in megabytes
```

```python
df["Category_cat"] = df["Category_str"].astype("category")
```

```python
df
```

```python
df.memory_usage(deep=True, index=False)
```

```python
1000462 / 1e6
```

```python
str(np.round(58 / 1.000462, decimals=2))
```

```python
m = 5_000_000
df = pd.DataFrame(
    {
        "String": np.random.choice(list("ABCDE"), size=m),
        "Categorical": pd.Categorical(np.random.choice(list("ABCDE"), size=m)),
    }
)
```

```python
df
```

```python
df.memory_usage(index=False, deep=True)
```

```python
%timeit df["String"].value_counts()
```

```python
%timeit df["Categorical"].value_counts()
```

```python
print(f"Using Categorical data is {154 / 20.9: .2f} times faster than raw string data")
```

### Encoding Categorical Data

```python
# One-Hot encoding
df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red", "Green"],
    "Size": ["Medium", "Large", "Small", "Medium", "Small"]
})
```

```python
df
```

```python
df_encoded = pd.get_dummies(df, columns=["Color"])
```

```python
df_encoded
```

```python
pd.get_dummies(df, columns=["Color"], prefix="C", prefix_sep="-")
```

```python
# ordinal encoding
df = pd.DataFrame({
    "Size": ["Small", "Large", "Medium", "Large", "Small"],
    "Quality": ["Low", "Average", "High", "Low", "High"],
})

df
```

```python
df["Size Encoded"] = pd.Categorical(df["Size"], categories=["Small", "Medium", "Large"], ordered=True).codes
df["Quality Encoded"] = pd.Categorical(df["Quality"], categories=["Low", "Average", "High"], ordered=True).codes
```
