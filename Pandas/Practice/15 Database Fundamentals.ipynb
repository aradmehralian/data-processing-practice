### Relational Database Concepts

- **Tables** (also called relations) represent entities or concepts (e.g., customers, orders).
- **Rows** (or records) represent individual instances of that entity.
- **Columns** (or fields) represent attributes of the entity.

````python
# DataFrames or Tables
customers
````

````python
orders["order_date"]
````

- **Primary Keys**: A column (or set of columns) that uniquely identifies each row in a table.
- **Foreign Keys**: A column that refers to the primary key of another table, establishing a relationship between the tables

````python
print("Primary Key in Customers Table:")
customers["customer_id"]
````

````python
print("Primary Key in Orders Table:")
orders["order_id"]
````

````python
print("Foreign Key in Orders Table:")
orders["customer_id"] # connects to customers table
````

Relationships Between Tables:

1. **One-to-One (1:1)**: Each record in Table A is related to exactly one record in Table B, and vice versa.

2. **One-to-Many (1:N)**: Each record in Table A can be related to multiple records in Table B, but each record in Table B is related to only one record in Table A.

3. **Many-to-Many (M:N)**: Multiple records in Table A can be related to multiple records in Table B, and vice versa.

There is a **One-to-Many** relationship between `customers` and `orders` tables. Each customer can have multiple orders, but each order belongs to exactly one customer.

````python
# number of orders per customer_id
orders["customer_id"].value_counts()
````

````python
pd.merge(customers, orders, on="customer_id", how="outer")
````

### Joining Tables

````python
# Adding a customer with no orders and an order with no matching customer
customers.loc[len(customers)] = {'customer_id': 5, 'name': 'Dua Lipa', 'email': 'dua@example.com'}
orders.loc[len(orders)] = {'order_id': 106, 'customer_id': 6, 'product': 'Widget D', 'quantity': 1, 'order_date': '2025-05-06'}
````

````python
customers
````

````python
orders
````

````python
# inner join
customers.merge(orders, on="customer_id", how="inner")
````

````python
# left join
pd.merge(customers, orders, on="customer_id", how="left")
````

````python
# right join
customers.merge(orders, on="customer_id", how="right")
````

````python
# full outer join
pd.merge(customers, orders, on="customer_id", how="outer")
````
