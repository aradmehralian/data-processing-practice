### Datetime Indexing and Slicing

```python
dates = pd.date_range(start="2025-01-01", end="2025-12-31", freq="B")
dates
```

```python
df = pd.DataFrame({"Value": np.random.randn(len(dates))}, index=dates)
df
```

```python
df.loc["2025-06-25"]
```

```python
# partial string indexing
df.loc["2025-08"]
```

```python
df.loc["2025-01":"2025-03"]
```

```python
df.loc["2025-12-10":]
```

```python
df.tail(6)
```

```python
df.loc[:"2025-04-28"]
```

```python
df.loc["2025-01":"2025-06"]
```

```python
df[df.index.day_name() == "Tuesday"]
```

```python
df[df.index.month == 4]
```

```python
df[(df.index.day_name() == "Friday") & (df.index.quarter == 2)]
```

```python
df[df.index.day > 16]
```

```python
df[(df.index.month == 8) & (df.index.day_name() == "Thursday") & (df["Value"] > 0)]
```

```python
df.resample("ME").max()
```
