```python
# 7
def vectorized_max(x: np.ndarray, y: np.ndarray) -> np.ndarray:
    return np.where(x >= y, x, y)
```

```python
def maxx(x, y):
    return x if x >= y else y
```

```python
vectorized_maxx = np.vectorize(maxx)
```

```python
a
```

```python
b
```

```python
vectorized_maxx(a, b)
```

```python
vectorized_max(a, b)
```

```python
np.maximum(a, b)
```

```python
%timeit vectorized_max(a, b)
```

```python
%timeit np.maximum(a, b)
```

```python
# 10
arr = np.random.randint(1, 100, size=20)
```

```python
def rolling_mean(arr: np.ndarray, window_size: int = 3) -> np.ndarray:
    window = np.ones(shape=window_size) / window_size
    return np.convolve(arr, window, mode="valid")
```

```python
rolling_mean(arr, 5)
```

```python
result = rolling_mean(np.arange(1, 11))
```

```python
result
```

```python
len(result)
```
