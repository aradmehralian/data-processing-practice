### Handling Mixed Categorical Data

```python
df_titanic.loc[0, "embarked"] = "Southampton"
df_titanic.loc[1, "embarked"] = "Cherbourg"
df_titanic.loc[2, "embarked"] = np.nan
```

```python
df_titanic["embarked"].unique()
```

```python
def clean_categorical_column(series):
    mapping = {"Southampton": "S", "Cherbourg": "C", "Queenstown": "Q"}

    cleaned_series = series.map(lambda x: mapping.get(x, x)).astype("category")
    return cleaned_series
```

```python
clean_data = clean_categorical_column(df_titanic["embarked"])
clean_data
```

```python
clean_data.unique()
```

### Handling Mixed Data Types

```python
df_taxis["pickup"] = "1912-04-10"
df_taxis.loc[0, "pickup"] = "10/04/1912"
df_taxis.loc[1, "pickup"] = "April 10, 1912"
df_taxis.loc[2, "pickup"] = np.nan
```

```python
def convert_to_iso(date_string: str):
    parsed_date = dateparser.parse(date_string)

    if parsed_date is None:
        return "Invalid date format"

    iso_date = parsed_date.isoformat()
    return iso_date
```

```python
converted_date = convert_to_iso("July 26th, 2026")
converted_date
```

```python
type(converted_date)
```

```python
def clean_date_column(series):
    cleaned_series = series.apply(lambda x: convert_to_iso(str(x)))
    return cleaned_series.astype("datetime64[s]")
```

```python
clean_date_column(df_taxis["pickup"])
```
