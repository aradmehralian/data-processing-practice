## Introduction to Exploratory Data Analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_iris = sns.load_dataset("iris")
```

```python
df_iris.columns
```

```python
_ = sns.pairplot(data=df_iris, height=1.5, hue="species")
```

```python
corr_mat = df_iris.drop(columns=["species"]).corr()
```

```python
corr_mat
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(ax=ax, data=corr_mat, annot=True, cmap="coolwarm")
ax.set_title("Correlation Heatmap of Iris Dataset", fontweight="bold")

fig.tight_layout()
```

```python
df_titanic = sns.load_dataset("titanic")
```

```python
missing_data_titanic = df_titanic.isnull().sum()
```

```python
print("Number of missing values in Titanic dataset:")
print(missing_data_titanic[missing_data_titanic > 0])
```

```python
fig, ax = plt.subplots(figsize=(10, 7))

sns.histplot(ax=ax, data=df_iris, x="sepal_length", hue="species", kde=True)
_ = ax.set_title("Distribution of Sepal Length by Species", fontweight="bold")
```

```python
fig, ax = plt.subplots(figsize=(10, 7))

sns.boxplot(ax=ax, data=df_iris, y="petal_length", x="species")
ax.set_title("Petal Length Distribution by Species", fontweight="bold")

fig.tight_layout()
```

```python
df_tips = sns.load_dataset("tips")
```

```python
mean_total_bill = df_tips["total_bill"].mean()
print(f"Mean Total Bill: {mean_total_bill:.2f}")
```

```python
median_total_bill = df_tips["total_bill"].median()
print(f"Median Total Bill: {median_total_bill:.2f}")
```

```python
sd_total_bill = df_tips["total_bill"].std()
print(f"Standard Deviation of Total Bill: {sd_total_bill:.2f}")
```

```python
range_total_bill = df_tips["total_bill"].max() - df_tips["total_bill"].min()
print(f"Range of Total Bill: {range_total_bill}")
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

sns.scatterplot(data=df_tips, ax=ax, x="total_bill", y="tip", hue="time", s=55)
ax.set_title("Relationship between Total Bill and Tip", fontweight="bold")

fig.tight_layout()
```

```python
day_summary = df_tips.groupby("day").agg({
    "total_bill": ["mean", "median", "count"],
    "tip": ["mean", "median"]
})
```

```python
round(day_summary, ndigits=2)
```

```python
df_flights = sns.load_dataset("flights")
```

```python
df_flights.columns
```

```python
fig, ax = plt.subplots(figsize=(11, 8))

sns.lineplot(ax=ax, data=df_flights, x="year", y="passengers", hue="month")
ax.set_title("Monthly Passenger Numbers over Time", fontweight="bold")

fig.tight_layout()
```

```python
df_tips["tip_percentage"] = (df_tips["tip"] / df_tips["total_bill"]) * 100
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

sns.boxplot(ax=ax, data=df_tips, x="day", y="tip_percentage", color="salmon")
ax.set_title("Tip Percentage by Day")

fig.tight_layout()
```
